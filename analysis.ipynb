# E-Commerce Customer Behavior Analysis

## 1. Project Overview & Business Objectives
This project focuses on analyzing customer behavior in an e-commerce setting to drive retention, optimize category strategy, and improve delivery experience.

### Key Analytical Focus Areas:
- **Average Order Value (AOV)** analysis across customer segments
- **Delivery Rating & Service Performance** evaluation
- **Product Category Distribution** and purchasing patterns

## 2. Environment Setup & Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set style for visualizations
sns.set_theme(style="whitegrid")
plt.rcParams['font.size'] = 11

# Ensure images directory exists for saving plots
os.makedirs('images', exist_ok=True)

# Load dataset
df = pd.read_csv('ecommerce_customer_behavior_dataset.csv')
df.head()
```

## 3. Exploratory Data Analysis (EDA)
Understanding data types, missing values, and summary statistics.

```python
# Dataset information and summary statistics
print("Dataset Info:")
df.info()

print("\nSummary Statistics:")
display(df.describe())
```

## 4. Visualizations & Insights

### 4.1 Average Order Value (AOV) Distribution
Analyzing spending patterns to identify core customer purchase power.

```python
# Visualizing Average Order Value
plt.figure(figsize=(9, 5))
sns.histplot(df['Total Spend'] if 'Total Spend' in df.columns else df.iloc[:, 1], kde=True, color='#2b5c8f')
plt.title('Distribution of Customer Spend / Order Value', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Order Value / Spend')
plt.ylabel('Customer Count')
plt.tight_layout()
plt.savefig('images/average_order_value.png', dpi=300)
plt.show()
```

### 4.2 Product Category Popularity
Evaluating order distribution across different categories.

```python
# Visualizing Product Category Distribution
plt.figure(figsize=(10, 5))
if 'Category' in df.columns:
    category_col = 'Category'
elif 'Product Category' in df.columns:
    category_col = 'Product Category'
else:
    category_col = df.select_dtypes(include=['object']).columns[0]

sns.countplot(data=df, y=category_col, palette='Blues_r', order=df[category_col].value_counts().index)
plt.title('Orders by Product Category', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Orders')
plt.ylabel('Category')
plt.tight_layout()
plt.savefig('images/product_category.png', dpi=300)
plt.show()
```

### 4.3 Delivery & Customer Satisfaction Ratings
Assessing delivery quality and customer feedback metrics.

```python
# Visualizing Delivery / Customer Ratings
plt.figure(figsize=(8, 5))
rating_col = [col for col in df.columns if 'rating' in col.lower() or 'score' in col.lower()]
rating_col = rating_col[0] if rating_col else df.select_dtypes(include=[np.number]).columns[-1]

sns.boxplot(data=df, x=rating_col, color='#4caf50')
plt.title('Customer Rating Distribution', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Rating / Score')
plt.tight_layout()
plt.savefig('images/delivery_rating.png', dpi=300)
plt.show()
```

## 5. Conclusions & Strategic Recommendations
- **Retention Strategy:** Target high AOV customer segments with personalized loyalty offers.
- **Category Optimization:** Focus marketing and inventory management on top-performing product categories.
- **Service Standards:** Maintain service quality based on delivery feedback scores to reduce churn.
